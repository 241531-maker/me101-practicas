```python
import pandas as pd
import numpy as np

# Completa: crea el diccionario con al menos 6 estudiantes y al menos 2 np.nan
datos = {
    "nombre": ["Ana", "Luis", "María", "Carlos", "Sofía", "Diego"],
    "curso": ["Estadística", "Estadística", "Cálculo", "Cálculo", "Álgebra", "Álgebra"],
    "nota": [15.5, 12.0, np.nan, 17.0, 14.5, 10.5],          # incluye al menos un np.nan aquí
    "asistencia_pct": [95, 80, 88, np.nan, 100, 70]          # o aquí
}

df_original = pd.DataFrame(datos)

# Completa: guarda como CSV
df_original.to_csv("estudiantes.csv", index=False)
```

```python
# Completa: lee el CSV
df = pd.read_csv("estudiantes.csv")

# Completa: muestra shape y cantidad de NaN por columna
print("Forma (filas, columnas):", df.shape)
print(df.isna().sum())
```

```python
# Completa: elimina filas con nota faltante
df_limpio = df.dropna(subset=["nota"])

# Completa: rellena asistencia_pct faltante con el promedio
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(df_limpio["asistencia_pct"].mean())

# Completa: verifica que no queden valores faltantes
print(df_limpio.isna().sum())
print("¿Quedan NaN?:", df_limpio.isna().any().any())   # debe dar False
```

```python
# Completa: promedio de nota agrupado por curso
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()
print(promedio_por_curso)
```

```python
# Guarda el DataFrame limpio
df_limpio.to_csv("estudiantes_limpio2.csv", index=False)
```
