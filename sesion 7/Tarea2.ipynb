```python
from PIL import Image, ImageEnhance, ImageFilter

# Reabrir la imagen de la Tarea 1
img2 = Image.open("mi_imagen.png")

# Completa: dos factores de brillo distintos
enh = ImageEnhance.Brightness(img2)
img_oscura = enh.enhance(0.5)    # factor menor a 1.0
img_clara  = enh.enhance(2.5)    # factor mayor a 2.0
img_oscura.save("mi_imagen_brillo_0_5.png")
img_clara.save("mi_imagen_brillo_2_5.png")

coord = (30, 30)   # pixel del rectángulo rojo (color saturado)
antes   = img2.getpixel(coord)
despues = img_clara.getpixel(coord)
print("ANTES  :", antes)
print("DESPUÉS:", despues)
print("¿Canales recortados (0 o 255)?:", [c for c in despues if c in (0, 255)])
```

```python
# Completa: aplica y guarda ambos filtros
img_blur  = img2.filter(ImageFilter.GaussianBlur(radius=3))
img_edges = img2.filter(ImageFilter.FIND_EDGES)
img_blur.save("mi_imagen_blur.png")
img_edges.save("mi_imagen_edges.png")
print("Blur:", img_blur.size, "| Edges:", img_edges.size)
```

```python
import numpy as np
from skimage import color, filters

# Completa: array, rgb2gray, comparacion con convert('L')
arr = np.array(img2)
print("shape:", arr.shape, "dtype:", arr.dtype)

gris_sk = color.rgb2gray(arr)      # valores entre 0 y 1
x, y = 100, 20                     # pixel de fondo de la Tarea 1
val_sk  = gris_sk[y, x] * 255      # en NumPy el orden es [fila, columna] = [y, x]
val_pil = img2.convert("L").getpixel((x, y))

print("scikit-image rgb2gray*255:", val_sk)
print("Pillow convert('L')      :", val_pil)
```

```python
# Completa: threshold_otsu y binarizacion
umbral  = filters.threshold_otsu(gris_sk)
binaria = gris_sk > umbral

print("Umbral de Otsu:", umbral)
print("Proporción de True:", binaria.mean())
```
