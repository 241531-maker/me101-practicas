```python
from PIL import Image, ImageDraw

# Completa: crea tu imagen con fondo y al menos 2 figuras
img = Image.new("RGB", (120, 120), (255, 244, 200))   
draw = ImageDraw.Draw(img)
draw.rectangle([10, 10, 60, 60], fill=(220, 50, 50))                 # figura 1
draw.ellipse([60, 60, 110, 110], fill=(40, 90, 200))                 # figura 2
draw.line([10, 110, 50, 70], fill=(30, 140, 60), width=4)            # figura 3

print(img.size, img.mode)
img.save("mi_imagen.png")
```

```python
# Completa: reabrir y verificar pixeles clave
img2 = Image.open("mi_imagen.png")
print("format:", img2.format, "| size:", img2.size, "| mode:", img2.mode)

print("Fondo     (100, 20):", img2.getpixel((100, 20)))
print("Figura 1  (30, 30) :", img2.getpixel((30, 30)))
print("Figura 2  (85, 85) :", img2.getpixel((85, 85)))
```

```python
# Completa: las cuatro transformaciones
img_L   = img2.convert("L")
img_rot = img2.rotate(90, expand=True)
img_res = img2.resize((img2.width // 2, img2.height // 2))
img_flp = img2.transpose(Image.FLIP_TOP_BOTTOM)

print("convert('L')   :", img_L.size, img_L.mode)
print("rotate(90)     :", img_rot.size)
print("resize (mitad) :", img_res.size)
print("flip vertical  :", img_flp.size)
```
