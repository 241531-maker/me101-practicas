```python
import numpy as np
import matplotlib.pyplot as plt

# Completa: construye el degradado de 5x5
gris = np.tile(np.linspace(0, 255, 5), (5, 1))

# Completa: imprime shape, dtype y mean()
print(gris.shape)   
print(gris.dtype) 
print(gris.mean())  
```

```python
# Completa: visualiza y guarda la imagen
plt.imshow(gris, cmap="gray")
plt.savefig("gris_tema6.png")
plt.show()
```

```python
# Completa: recorte de la mitad derecha
recorte = gris[:, -3:]
print(recorte)
```

```python
# Completa: construye la imagen RGB de dos mitades
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [200, 50, 50]    # mitad izquierda: rojo (R,G,B)
rgb[:, 2:4] = [50, 100, 200]   # mitad derecha: azul

print(rgb.shape)               # (4, 4, 3)
print("Pixel [0,0]:", rgb[0, 0])   # [200  50  50]
print("Pixel [0,3]:", rgb[0, 3])   # [ 50 100 200]
```

```python
# Completa: extrae cada canal
R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

print("R:\n", R)
print("G:\n", G)
print("B:\n", B)
```

```python
# Completa: gris_promedio y gris_luminosa
# Se convierte a float para evitar el desbordamiento de uint8 al sumar
R, G, B = R.astype(float), G.astype(float), B.astype(float)

gris_promedio = (R + G + B) / 3
gris_luminosa = 0.299 * R + 0.587 * G + 0.114 * B

print("Promedio:\n", gris_promedio)
print("Luminosa:\n", gris_luminosa)
```

```python
# Completa: binarización con tu propio umbral
umbral = 95
binaria = (gris_luminosa > umbral).astype(np.uint8)

print(binaria)
```
